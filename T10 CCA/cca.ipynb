```python
cols=[var for var in df.columns if df[var].isnull().mean() < 0.05 and df[var].isnull().mean()>0]
cols
```

```python
df[cols].sample(5)
```

```python
len(df[cols].dropna())/len(df)
```

```python
new_df=df[cols].dropna()
df.shape, new_df.shape
```

#### Checking for Numerical Cols

```python
new_df.hist(figsize=(10,10))
plt.show()
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

df['training_hours'].hist(bins=50, ax=ax, density=True, color='blue', alpha=0.5)
new_df['training_hours'].hist(bins=50, ax=ax, density=True, color='yellow', alpha=0.5)
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

df['city_development_index'].hist(bins=50, ax=ax, density=True, color='blue')
new_df['city_development_index'].hist(bins=50, ax=ax, density=True, color='yellow')
```

```python
fig=plt.figure()
ax=fig.add_subplot(111)

df['experience'].hist(bins=50, ax=ax, density=True, color='blue')
new_df['experience'].hist(bins=50, ax=ax, density=True, color='yellow')
```

```python
temp=pd.concat([df["enrolled_university"].value_counts() / len(df), new_df["enrolled_university"].value_counts()/ len(new_df)], axis=1)

temp.columns=['original', 'cca']
temp
```

```python
temp=pd.concat([df["education_level"].value_counts() / len(df), new_df["education_level"].value_counts()/ len(new_df)], axis=1)

temp.columns=['original', 'cca']
temp
```
